#1. Introduction

Nearby locations often exhibited similar levels of **mobility intensity** while supporting very different levels of **local economic intensity**. We used **mobility intensity** to refer to the volume of visitation observed in a place. We used **local economic intensity** to refer to the concentration of commerce, services, circulation, and built-environment activity associated with ongoing economic presence. This mismatch created a practical problem: traffic alone did not reveal whether a place supported repeated access, stable visitation, and stronger local economic conditions.

This problem affected real decisions. Firms choosing where to locate, entrepreneurs evaluating local opportunity, and planners targeting revitalization often relied on visible activity or static territorial proxies such as density, land use, and socioeconomic composition. These measures described places in broad terms, but they did not show how people actually used places through return, dwell time, and temporal regularity. As a result, places with similar traffic often differed in the underlying conditions that sustained local activity.

We addressed this gap by treating mobility not only as volume, but as a structured behavioral signal. We conceptualized **market infrastructure** as the latent spatial organization of access conditions that supported repeated visitation, stable exposure, and sustained local interaction. We used the term carefully. We did not use it to denote market outcomes directly, and we did not restrict it to visible physical infrastructure. Instead, we used it to describe an underlying spatial condition inferred from how mobility behavior was organized across places.

This distinction mattered because two locations could exhibit similar mobility intensity while differing sharply in visitor diversity, repeat visitation, dwell time, and temporal stability. Such differences indicated whether mobility mainly passed through space or unfolded in patterns more consistent with recurring access and persistent local activation. We therefore asked:

**Can a latent form of market infrastructure be identified from spatial patterns of human mobility, and can it explain spatial differences in local economic intensity?**

To answer this question, we developed a Spatial AI framework that combined tract-level mobility indicators, dimensionality reduction, spatial regime detection, external economic validation, and multiscale reconstruction. We constructed the **Market Infrastructure Index (MII)** from the dominant covariance structure of mobility behavior, analyzed how this index varied across Brazil’s census tracts, identified **mobility infrastructure regimes**, evaluated their association with **VIIRS nighttime light intensity**, and implemented **mass-preserving downscaling** in São Paulo to recover sub-tract variation at 100-meter resolution.

This notebook showed that mobility data could do more than map flows. We used it to recover a latent spatial structure associated with repeated access, stable visitation, and stronger local economic intensity. In doing so, we moved beyond traffic counts and showed why places with similar mobility intensity could sustain very different local conditions.

```python
# ============================================================
# SECTION 2 — Setup and helper functions
# ============================================================

# If you are running this notebook in a clean environment,
# uncomment the next line and install the required libraries:
# %pip install -q geopandas rasterio pyogrio

from pathlib import Path
from urllib.request import urlretrieve
from zipfile import ZipFile
import warnings

import geopandas as gpd
import pandas as pd
import rasterio
from IPython.display import display

warnings.filterwarnings("ignore", category=FutureWarning)

pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 180)
pd.set_option("display.max_colwidth", 120)


def print_banner(title: str) -> None:
    line = "=" * 72
    print(f"\n{line}\n{title}\n{line}")


def download_file(url: str, output_path: Path) -> Path:
    """
    Download a file only if it does not already exist locally.
    """
    output_path = Path(output_path)
    output_path.parent.mkdir(parents=True, exist_ok=True)

    if output_path.exists() and output_path.stat().st_size > 0:
        print(f"Using cached file: {output_path.name}")
        return output_path

    print(f"Downloading: {output_path.name}")
    urlretrieve(url, output_path)
    print(f"Saved to: {output_path}")
    return output_path


def unzip_file(zip_path: Path, extract_dir: Path) -> Path:
    """
    Extract a ZIP archive only once.
    """
    zip_path = Path(zip_path)
    extract_dir = Path(extract_dir)
    extract_dir.mkdir(parents=True, exist_ok=True)

    sentinel = extract_dir / ".unzipped"
    if sentinel.exists():
        print(f"Using cached extraction: {extract_dir}")
        return extract_dir

    print(f"Extracting: {zip_path.name}")
    with ZipFile(zip_path, "r") as zf:
        zf.extractall(extract_dir)

    sentinel.touch()
    print(f"Extracted to: {extract_dir}")
    return extract_dir


def find_single_file(root_dir: Path, pattern: str) -> Path:
    """
    Return the first file matching a pattern inside a directory tree.
    """
    matches = sorted(Path(root_dir).rglob(pattern))
    if not matches:
        raise FileNotFoundError(f"No files matched pattern '{pattern}' in {root_dir}")
    return matches[0]


def summarize_viirs_tile(tile_path: Path) -> dict:
    """
    Return a compact metadata summary for a VIIRS raster tile.
    """
    with rasterio.open(tile_path) as src:
        arr = src.read(1, masked=True)
        return {
            "tile": tile_path.name,
            "shape": src.shape,
            "crs": str(src.crs),
            "bounds": str(src.bounds),
            "min": float(arr.min()),
            "max": float(arr.max()),
            "mean": float(arr.mean()),
        }
```

# 3. From Mobility Signals to the Market Infrastructure Index

This section showed how we moved from tract-level mobility signals to the **Market Infrastructure Index (MII)**. We began with the seven standardized indicators embedded in the shared GeoPackage, examined their correlation structure, evaluated whether they reflected a dominant latent dimension, and verified that this dominant component aligned with the stored MII.

This sequence mattered because the MII was not intended to operate as a black-box score. It was designed to summarize a structured pattern of mobility behavior associated with repeated access, stable visitation, and local reactivation.

```python
# ============================================================
# SECTION 3 — Imports and mobility indicator configuration
# ============================================================

# If needed in a clean environment, uncomment the next line:
# %pip install -q seaborn scikit-learn

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA

sns.set_theme(style="whitegrid", context="notebook")

mobility_vars = [
    "z_log1p_visits_A4",
    "z_log1p_unique_A4",
    "z_log1p_repeat_visitors_A4",
    "z_log1p_new_visitors_A4",
    "z_dwell_time_mins",
    "z_stability_unique_week_cv_A4",
    "z_stability_visits_week_cv_A4",
]

mobility_labels = {
    "z_log1p_visits_A4": "Visit volume",
    "z_log1p_unique_A4": "Unique visitors",
    "z_log1p_repeat_visitors_A4": "Repeat visitors",
    "z_log1p_new_visitors_A4": "New visitors",
    "z_dwell_time_mins": "Dwell time",
    "z_stability_unique_week_cv_A4": "Stability (unique visitors)",
    "z_stability_visits_week_cv_A4": "Stability (visit counts)",
}

missing_mobility_vars = [var for var in mobility_vars if var not in gdf.columns]
if missing_mobility_vars:
    raise KeyError(f"Missing mobility variables in GeoPackage: {missing_mobility_vars}")
```

### 3.1 Mobility signals used in the framework

We characterized mobility behavior with seven standardized indicators capturing complementary dimensions of visitation activity: visit volume, unique visitor presence, inflow of new visitors, repeat-visitation intensity, average dwell time, temporal stability in unique visitors, and temporal stability in visit counts.

Because the shared GeoPackage already contained these fields in standardized form, we used them directly rather than reprocessing transformations that had already been embedded in the final asset.

```python
# ============================================================
# SECTION 3.1 — Availability and descriptive summary
# ============================================================

mobility_availability = pd.DataFrame(
    {
        "indicator": [mobility_labels[var] for var in mobility_vars],
        "column": mobility_vars,
        "non_missing_observations": [int(gdf[var].notna().sum()) for var in mobility_vars],
        "missing_observations": [int(gdf[var].isna().sum()) for var in mobility_vars],
    }
)

mobility_summary = (
    gdf[mobility_vars]
    .describe()
    .T[["count", "mean", "std", "min", "25%", "50%", "75%", "max"]]
    .round(2)
    .reset_index()
    .rename(columns={"index": "column"})
)
mobility_summary.insert(
    0,
    "indicator",
    mobility_summary["column"].map(mobility_labels),
)

print_banner("MOBILITY INDICATORS — AVAILABILITY")
display(mobility_availability)

print_banner("MOBILITY INDICATORS — DESCRIPTIVE SUMMARY")
display(mobility_summary)
```

The summary statistics confirmed substantial heterogeneity across census tracts. Because these indicators were already standardized, the reported values described relative differences in visitation intensity, recurrence, duration, and temporal regularity on a comparable scale.

### 3.2 Correlation structure of mobility behavior

We next examined whether the seven indicators moved together in a way consistent with a broader behavioral structure. If multiple indicators co-varied systematically, that pattern would suggest that mobility behavior was organized by a shared latent dimension rather than by isolated signals.

```python
# ============================================================
# SECTION 3.2 — Correlation structure
# ============================================================

corr = gdf[mobility_vars].corr()
corr_labeled = corr.rename(index=mobility_labels, columns=mobility_labels)

plt.figure(figsize=(9, 7))
sns.heatmap(
    corr_labeled,
    annot=True,
    fmt=".2f",
    cmap="RdBu_r",
    center=0,
    square=True,
    linewidths=0.5,
    cbar_kws={"shrink": 0.85, "label": "Correlation"},
)
plt.title("Correlation Structure of Mobility Indicators", fontsize=13, pad=12)
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

print_banner("CORRELATION MATRIX")
display(corr_labeled.round(2))
```

The correlation structure revealed a clear activity block linking visit volume, unique visitors, repeat visitors, and new visitors. Dwell time and the two stability indicators contributed distinct information, which was important because the framework aimed to capture more than traffic magnitude alone.

### 3.3 Evidence of a dominant latent dimension

We then evaluated whether the seven indicators exhibited a dominant common dimension. Because the shared GeoPackage already stored the indicators in standardized form, we estimated principal component analysis directly on those fields rather than standardizing them again.

```python
# ============================================================
# SECTION 3.3 — Principal component analysis
# ============================================================

X_pca = gdf[mobility_vars].dropna().copy()

pca = PCA()
pca.fit(X_pca.values)

pc_scores = pca.transform(X_pca.values)
stored_mii = gdf.loc[X_pca.index, "mii"].values

pc1_scores_raw = pc_scores[:, 0]
pc1_corr_raw = np.corrcoef(pc1_scores_raw, stored_mii)[0, 1]

pc1_sign = 1 if pc1_corr_raw >= 0 else -1
pc1_scores = pc1_sign * pc1_scores_raw
pc1_loadings = pc1_sign * pca.components_[0]
pc1_mii_corr = abs(pc1_corr_raw)

explained_variance_df = pd.DataFrame(
    {
        "component": [f"PC{i}" for i in range(1, len(mobility_vars) + 1)],
        "explained_variance_pct": np.round(100 * pca.explained_variance_ratio_, 2),
        "cumulative_variance_pct": np.round(100 * np.cumsum(pca.explained_variance_ratio_), 2),
    }
)

print_banner("PCA SUMMARY")
print(f"Observations used in PCA: {X_pca.shape[0]:,}")
print(f"Variables used in PCA: {X_pca.shape[1]}")
display(explained_variance_df)

plt.figure(figsize=(7.5, 5))
plt.plot(
    explained_variance_df["component"],
    explained_variance_df["explained_variance_pct"],
    marker="o",
    linewidth=2,
)
plt.ylabel("Explained Variance (%)")
plt.xlabel("Principal Component")
plt.title("Explained Variance of Mobility Indicators", fontsize=13, pad=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

The PCA results showed that the first component captured a large share of the total variance, while the remaining components explained substantially smaller shares. This pattern indicated that the seven indicators were not behaving as disconnected signals. Instead, they reflected a strong common dimension underlying mobility behavior across census tracts.

### 3.4 Interpreting the first principal component

To understand what the dominant component represented, we examined the loadings of the first principal component. These loadings showed how strongly each mobility indicator contributed to the latent dimension summarized by the MII.

```python
# ============================================================
# SECTION 3.4 — PC1 loadings
# ============================================================

pc1_loadings_df = pd.DataFrame(
    {
        "indicator": [mobility_labels[var] for var in mobility_vars],
        "column": mobility_vars,
        "pc1_loading": np.round(pc1_loadings, 4),
        "absolute_loading": np.round(np.abs(pc1_loadings), 4),
    }
).sort_values("pc1_loading")

print_banner("FIRST PRINCIPAL COMPONENT LOADINGS")
display(pc1_loadings_df)

plt.figure(figsize=(8, 5))
plt.barh(
    pc1_loadings_df["indicator"],
    pc1_loadings_df["pc1_loading"],
)
plt.axvline(0, color="black", linewidth=1)
plt.xlabel("PC1 Loading")
plt.title("Loadings of Mobility Indicators on the First Principal Component", fontsize=13, pad=12)
plt.tight_layout()
plt.show()
```

The first principal component was driven primarily by the coordinated alignment of activity-related signals, especially visit volume, unique visitors, repeat visitors, and new visitors. The stability indicators also contributed to this latent dimension, while dwell time captured a more distinct behavioral aspect. Taken together, these patterns supported the interpretation of PC1 as a structured dimension of repeated access and stable visitation rather than as a simple traffic count.

### 3.5 Consistency between the recomputed first component and the stored MII

Finally, we verified that the first principal component recomputed from the seven standardized indicators aligned with the **MII** stored in the shared GeoPackage. This step strengthened transparency by showing that the shared analytical asset preserved the same latent dimension used throughout the study.

```python
# ============================================================
# SECTION 3.5 — Consistency check between recomputed PC1 and stored MII
# ============================================================

alignment_df = pd.DataFrame(
    {
        "recomputed_pc1": pc1_scores,
        "stored_mii": stored_mii,
    },
    index=X_pca.index,
)

alignment_summary = pd.DataFrame(
    {
        "metric": [
            "observations_compared",
            "correlation_recomputed_pc1_vs_stored_mii",
            "mean_recomputed_pc1",
            "mean_stored_mii",
            "std_recomputed_pc1",
            "std_stored_mii",
        ],
        "value": [
            len(alignment_df),
            round(pc1_mii_corr, 6),
            round(alignment_df["recomputed_pc1"].mean(), 6),
            round(alignment_df["stored_mii"].mean(), 6),
            round(alignment_df["recomputed_pc1"].std(), 6),
            round(alignment_df["stored_mii"].std(), 6),
        ],
    }
)

print_banner("MII CONSISTENCY CHECK")
display(alignment_summary)

plt.figure(figsize=(6.8, 5.8))
hb = plt.hexbin(
    alignment_df["recomputed_pc1"],
    alignment_df["stored_mii"],
    gridsize=60,
    mincnt=1,
)
lims = [
    min(alignment_df["recomputed_pc1"].min(), alignment_df["stored_mii"].min()),
    max(alignment_df["recomputed_pc1"].max(), alignment_df["stored_mii"].max()),
]
plt.plot(lims, lims, linestyle="--", linewidth=1)
plt.xlabel("Recomputed PC1 Score")
plt.ylabel("Stored MII")
plt.title("Alignment Between Recomputed PC1 and Stored MII", fontsize=13, pad=12)
cb = plt.colorbar(hb)
cb.set_label("Number of Census Tracts")
plt.tight_layout()
plt.show()
```

The strong alignment between the recomputed first component and the stored MII showed that the shared GeoPackage preserved the same dominant latent dimension extracted from the seven standardized mobility indicators.

At this point, we had established three things. First, the mobility indicators represented complementary dimensions of visitation behavior. Second, their correlation structure was highly organized rather than diffuse. Third, this organization was dominated by a common latent dimension that corresponded closely to the Market Infrastructure Index used throughout the notebook.

# 4.Spatial Regime Discovery

The national latent structure identified in the previous section did not imply that mobility behavior was organized uniformly across space. Instead, the paper’s core argument was that the covariance structure of mobility could vary geographically, revealing distinct spatial environments that would remain hidden under a single national decomposition.

To recover this spatial heterogeneity, we estimated localized covariance signatures from the same seven standardized mobility indicators used to construct the MII. We then discretized those localized signatures into a finite set of **mobility infrastructure regimes**.

This step mattered because places could differ not only in the overall level of mobility-based infrastructure, but also in the **way mobility behavior was organized locally**. Regime discovery therefore moved the analysis beyond a single national latent dimension and toward a spatially differentiated view of mobility-based infrastructure.

```python
# ============================================================
# SECTION 4 — Imports and configuration
# ============================================================

# If needed in a clean environment, uncomment the next line:
# %pip install -q scikit-learn

from sklearn.cluster import KMeans
from sklearn.neighbors import NearestNeighbors
from sklearn.decomposition import PCA

regime_vars = mobility_vars.copy()

# Metric CRS used for neighborhood search and centroid-based distances
metric_crs = 5880

# Spatial regime configuration
target_focal_locations = 25000
k_neighbors = 200
n_regimes = 4
random_state = 42
```

### 4.1 Preparing the localized regime workflow

To keep the notebook fully runnable while preserving national coverage, we estimated localized covariance signatures at systematically distributed focal locations and then propagated the resulting regime labels to the nearest valid tract centroids.

This approach preserved the substantive logic of the framework while making the spatial regime step computationally tractable in notebook form.

```python
# ============================================================
# SECTION 4.1 — Prepare valid tract-level inputs in metric space
# ============================================================

gdf_metric = gdf.to_crs(metric_crs).copy()

gdf_regime_valid = gdf_metric.dropna(subset=regime_vars).copy()
gdf_regime_valid["centroid"] = gdf_regime_valid.geometry.centroid

coords_valid = np.column_stack(
    [
        gdf_regime_valid["centroid"].x.values,
        gdf_regime_valid["centroid"].y.values,
    ]
)

X_valid = gdf_regime_valid[regime_vars].to_numpy()

print_banner("SPATIAL REGIME WORKFLOW — VALID INPUTS")
print(f"Valid census tracts used for localized covariance analysis: {len(gdf_regime_valid):,}")
print(f"Mobility matrix shape: {X_valid.shape}")
print(f"Coordinate matrix shape: {coords_valid.shape}")
print(f"Metric CRS used for neighborhood search: EPSG:{metric_crs}")
```

```python
# ============================================================
# SECTION 4.1 — Select systematically distributed focal locations
# ============================================================

def select_focal_locations(coords: np.ndarray, target_n: int) -> np.ndarray:
    """
    Select spatially distributed focal locations using deterministic grid thinning.
    """
    if len(coords) <= target_n:
        return np.arange(len(coords))

    x = coords[:, 0]
    y = coords[:, 1]

    xmin, xmax = x.min(), x.max()
    ymin, ymax = y.min(), y.max()

    width = max(xmax - xmin, 1.0)
    height = max(ymax - ymin, 1.0)

    cell_size = np.sqrt((width * height) / target_n)
    cell_size = max(cell_size, 1.0)

    cell_x = np.floor((x - xmin) / cell_size).astype(np.int64)
    cell_y = np.floor((y - ymin) / cell_size).astype(np.int64)

    cell_df = pd.DataFrame({"cell_x": cell_x, "cell_y": cell_y})
    focal_idx = cell_df.drop_duplicates().index.to_numpy()

    if len(focal_idx) > target_n:
        focal_idx = focal_idx[np.linspace(0, len(focal_idx) - 1, target_n, dtype=int)]

    return np.sort(focal_idx)


focal_idx = select_focal_locations(coords_valid, target_focal_locations)
coords_focal = coords_valid[focal_idx]

focal_summary = pd.DataFrame(
    {
        "metric": [
            "valid_tracts",
            "target_focal_locations",
            "selected_focal_locations",
            "k_neighbors",
            "n_regimes",
        ],
        "value": [
            len(gdf_regime_valid),
            target_focal_locations,
            len(focal_idx),
            k_neighbors,
            n_regimes,
        ],
    }
)

print_banner("FOCAL LOCATION SELECTION")
display(focal_summary)
```

### 4.2 Localized covariance signatures

For each focal location, we identified a local neighborhood of nearby tracts and estimated a two-component PCA on the seven standardized mobility indicators. We then represented each focal location by:

- the local share of variance explained by the first two components, and
- the focal tract’s score on those two localized components.

Together, these quantities summarized how mobility behavior was organized in the focal neighborhood.

```python
# ============================================================
# SECTION 4.2 — Estimate localized PCA signatures
# ============================================================

# Build nearest-neighbor structure on all valid tract centroids
nn_all = NearestNeighbors(
    n_neighbors=k_neighbors,
    algorithm="ball_tree",
)
nn_all.fit(coords_valid)

_, neighbor_indices = nn_all.kneighbors(coords_focal)

# Recover global reference directions from Section 3
# If Section 3 was run immediately before, these objects already exist.
# We still define them defensively here for clarity.
global_pc1_vector = np.asarray(pc1_loadings)
global_pc2_vector = np.asarray(pca.components_[1])

local_signature_rows = []

for focal_position, tract_idx in enumerate(focal_idx):
    local_idx = neighbor_indices[focal_position]
    X_local = X_valid[local_idx]

    local_pca = PCA(n_components=2)
    local_pca.fit(X_local)

    local_pc1 = local_pca.components_[0].copy()
    local_pc2 = local_pca.components_[1].copy()

    # Align local component signs to the global PCA directions
    if np.dot(local_pc1, global_pc1_vector) < 0:
        local_pc1 *= -1

    if np.dot(local_pc2, global_pc2_vector) < 0:
        local_pc2 *= -1

    focal_values = X_valid[tract_idx].reshape(1, -1)
    local_scores = local_pca.transform(focal_values)[0]

    local_signature_rows.append(
        {
            "ct_id": gdf_regime_valid.iloc[tract_idx]["ct_id"],
            "focal_pc1_score": local_scores[0],
            "focal_pc2_score": local_scores[1],
            "pc1_share": local_pca.explained_variance_ratio_[0],
            "pc2_share": local_pca.explained_variance_ratio_[1],
        }
    )

focal_signatures = pd.DataFrame(local_signature_rows)

print_banner("LOCALIZED PCA SIGNATURES")
display(focal_signatures.head())
```

The localized PCA step produced a spatially varying representation of mobility organization. We next discretized that continuous field into a finite number of interpretable spatial regimes.

```python
# ============================================================
# SECTION 4.3 — Cluster localized signatures into regimes
# ============================================================

cluster_features = focal_signatures[["focal_pc1_score", "focal_pc2_score"]].to_numpy()

kmeans = KMeans(
    n_clusters=n_regimes,
    n_init=20,
    random_state=random_state,
)

raw_regimes = kmeans.fit_predict(cluster_features)
focal_signatures["raw_regime"] = raw_regimes

# Reorder regime labels deterministically by descending mean PC1 share
regime_order = (
    focal_signatures.groupby("raw_regime")["pc1_share"]
    .mean()
    .sort_values(ascending=False)
    .index
    .tolist()
)

regime_mapping = {old_label: new_label for new_label, old_label in enumerate(regime_order)}
focal_signatures["infrastructure_regime"] = focal_signatures["raw_regime"].map(regime_mapping)

regime_counts = (
    focal_signatures["infrastructure_regime"]
    .value_counts()
    .sort_index()
    .rename_axis("infrastructure_regime")
    .reset_index(name="n_focal_locations")
)

print_banner("REGIME DISCOVERY — FOCAL LOCATION COUNTS")
display(regime_counts)
```

### 4.3 Propagating regime labels to valid census tracts

Because the localized covariance structure was estimated at focal locations, we next propagated the resulting regime labels to the nearest valid census tract centroids. This yielded a tract-level regime layer that preserved national coverage for all tracts with complete mobility information.

```python
# ============================================================
# SECTION 4.3 — Assign each valid tract to its nearest focal regime
# ============================================================

nn_focal = NearestNeighbors(n_neighbors=1, algorithm="ball_tree")
nn_focal.fit(coords_focal)

_, nearest_focal_idx = nn_focal.kneighbors(coords_valid)

nearest_focal_idx = nearest_focal_idx.flatten()

gdf_regime_valid["focal_pc1_score"] = focal_signatures.iloc[nearest_focal_idx]["focal_pc1_score"].to_numpy()
gdf_regime_valid["focal_pc2_score"] = focal_signatures.iloc[nearest_focal_idx]["focal_pc2_score"].to_numpy()
gdf_regime_valid["pc1_share_local"] = focal_signatures.iloc[nearest_focal_idx]["pc1_share"].to_numpy()
gdf_regime_valid["pc2_share_local"] = focal_signatures.iloc[nearest_focal_idx]["pc2_share"].to_numpy()
gdf_regime_valid["infrastructure_regime"] = focal_signatures.iloc[nearest_focal_idx]["infrastructure_regime"].to_numpy()

# Merge regime outputs back to the original GeoDataFrame
regime_output_columns = [
    "ct_id",
    "focal_pc1_score",
    "focal_pc2_score",
    "pc1_share_local",
    "pc2_share_local",
    "infrastructure_regime",
]

gdf = gdf.merge(
    gdf_regime_valid[regime_output_columns],
    on="ct_id",
    how="left",
)

tract_regime_counts = (
    gdf["infrastructure_regime"]
    .value_counts(dropna=False)
    .sort_index()
    .rename_axis("infrastructure_regime")
    .reset_index(name="n_census_tracts")
)

print_banner("TRACT-LEVEL REGIME COUNTS")
display(tract_regime_counts)
```

# 5. External Economic Validation

The mobility infrastructure regimes identified in the previous section were not intended to remain purely behavioral categories. In this section, we evaluated whether those regimes also corresponded to distinct economic environments.

To do so, we used **VIIRS nighttime light intensity** as an external proxy for local economic intensity. We first extracted tract-level nighttime light values from the annual VIIRS archive and then compared their distribution across mobility infrastructure regimes. We next estimated **regime-specific mobility–economic response profiles** using cubic B-splines, allowing the association between mobility intensity and local economic intensity to vary in both magnitude and geometry across regimes.

This step was important because it moved the framework from structural differentiation alone to **external economic validation**.

```python
# ============================================================
# SECTION 5 — Imports and configuration
# ============================================================

# If needed in a clean environment, uncomment the next line:
# %pip install -q rasterstats patsy scikit-learn shapely

from pathlib import Path
from shapely.geometry import box

from rasterstats import zonal_stats
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import SplineTransformer

# VIIRS / validation configuration
nightlight_field = "nightlight_mean"
mobility_intensity_var = "z_log1p_visits_A4"   # tract-level mobility intensity proxy
viirs_crs = 4326

# Spline / bootstrap configuration
n_knots = 5
spline_degree = 3
bootstrap_iterations = 100
grid_points = 150
response_random_state = 42
```

### 5.2 Extracting tract-level nighttime light intensity

Instead of creating a full national raster mosaic in memory, we processed the VIIRS archive tile by tile. For each tile, we identified the census tracts intersecting its spatial extent, computed zonal statistics locally, and accumulated the tract-level results across tiles.

This approach was more memory efficient and more appropriate for a reproducible notebook workflow.

```python
# ============================================================
# SECTION 5.2 — Tile-by-tile VIIRS extraction
# ============================================================

if nightlight_field in gdf_viirs.columns:
    print_banner("NIGHTLIGHT EXTRACTION")
    print(f"Column '{nightlight_field}' already exists. Reusing existing values.")
else:
    # Prepare accumulation vectors
    gdf_viirs["viirs_sum"] = 0.0
    gdf_viirs["viirs_count"] = 0.0

    # Build spatial index once
    spatial_index = gdf_viirs.sindex

    for tile_path in viirs_tiles:
        with rasterio.open(tile_path) as src:
            tile_bounds = src.bounds
            tile_bbox = box(tile_bounds.left, tile_bounds.bottom, tile_bounds.right, tile_bounds.top)

            intersecting_idx = list(spatial_index.query(tile_bbox, predicate="intersects"))
            if len(intersecting_idx) == 0:
                continue

            tract_subset = gdf_viirs.iloc[intersecting_idx].copy()

            zs = zonal_stats(
                tract_subset,
                tile_path,
                stats=["sum", "count"],
                nodata=-1,
                all_touched=False,
            )

            tile_sum = np.array([z["sum"] if z["sum"] is not None else 0.0 for z in zs], dtype=float)
            tile_count = np.array([z["count"] if z["count"] is not None else 0.0 for z in zs], dtype=float)

            gdf_viirs.loc[tract_subset.index, "viirs_sum"] += tile_sum
            gdf_viirs.loc[tract_subset.index, "viirs_count"] += tile_count

    gdf_viirs[nightlight_field] = np.where(
        gdf_viirs["viirs_count"] > 0,
        gdf_viirs["viirs_sum"] / gdf_viirs["viirs_count"],
        np.nan,
    )

    # Push extracted values back to the main GeoDataFrame
    gdf = gdf.merge(
        gdf_viirs[["ct_id", nightlight_field]],
        on="ct_id",
        how="left",
        suffixes=("", "_new"),
    )

    if f"{nightlight_field}_new" in gdf.columns:
        gdf[nightlight_field] = gdf[f"{nightlight_field}_new"].combine_first(gdf[nightlight_field])
        gdf.drop(columns=[f"{nightlight_field}_new"], inplace=True)

print_banner("TRACT-LEVEL NIGHTLIGHT EXTRACTION")
print(gdf[nightlight_field].describe().round(3))
```

### 5.4 Regime-specific mobility–economic response profiles

The paper’s economic validation step did not stop at comparing average nighttime luminosity across regimes. It also evaluated whether the relationship between **mobility intensity** and **local economic intensity** varied across regimes.

To remain consistent with the conceptual definition used throughout the notebook, we represented **mobility intensity** with the standardized log-transformed visit-volume measure `z_log1p_visits_A4`. We then estimated cubic B-spline response functions separately for each regime, allowing the geometry of the mobility–economic relationship to differ across localized covariance environments.

```python
# ============================================================
# SECTION 5.4 — Helper function for spline-based response estimation
# ============================================================

def fit_regime_spline(
    x: np.ndarray,
    y: np.ndarray,
    n_knots: int = 5,
    degree: int = 3,
    bootstrap_iterations: int = 100,
    grid_points: int = 150,
    random_state: int = 42,
) -> dict:
    """
    Fit a cubic B-spline response curve with bootstrap uncertainty bands.
    """
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    valid_mask = np.isfinite(x) & np.isfinite(y)
    x = x[valid_mask]
    y = y[valid_mask]

    x_low, x_high = np.quantile(x, [0.01, 0.99])
    keep_mask = (x >= x_low) & (x <= x_high)

    x_fit = x[keep_mask]
    y_fit = y[keep_mask]

    if len(x_fit) < 100:
        raise ValueError("Not enough observations to estimate a stable spline.")

    spline = SplineTransformer(
        n_knots=n_knots,
        degree=degree,
        include_bias=False,
    )
    X_basis = spline.fit_transform(x_fit.reshape(-1, 1))

    model = LinearRegression()
    model.fit(X_basis, y_fit)

    x_grid = np.linspace(x_low, x_high, grid_points)
    X_grid = spline.transform(x_grid.reshape(-1, 1))
    y_pred = model.predict(X_grid)

    rng = np.random.default_rng(random_state)
    boot_predictions = []

    n = len(x_fit)
    for _ in range(bootstrap_iterations):
        sample_idx = rng.integers(0, n, size=n)
        x_boot = x_fit[sample_idx]
        y_boot = y_fit[sample_idx]

        X_boot = spline.transform(x_boot.reshape(-1, 1))
        boot_model = LinearRegression()
        boot_model.fit(X_boot, y_boot)
        boot_predictions.append(boot_model.predict(X_grid))

    boot_predictions = np.vstack(boot_predictions)
    pred_low = np.percentile(boot_predictions, 2.5, axis=0)
    pred_high = np.percentile(boot_predictions, 97.5, axis=0)

    elasticity = np.gradient(y_pred, x_grid)
    curvature = np.gradient(elasticity, x_grid)

    idx_p10 = np.argmin(np.abs(x_grid - np.quantile(x_grid, 0.10)))
    idx_p90 = np.argmin(np.abs(x_grid - np.quantile(x_grid, 0.90)))

    return {
        "x_grid": x_grid,
        "y_pred": y_pred,
        "pred_low": pred_low,
        "pred_high": pred_high,
        "elasticity": elasticity,
        "curvature": curvature,
        "elasticity_p10": elasticity[idx_p10],
        "elasticity_mean": elasticity.mean(),
        "elasticity_p90": elasticity[idx_p90],
        "p90_band_width": pred_high[idx_p90] - pred_low[idx_p90],
        "max_abs_curvature": np.abs(curvature).max(),
        "n_used": len(x_fit),
    }
```

```python
# ============================================================
# SECTION 5.4 — Estimate regime-specific response curves
# ============================================================

response_df = gdf.dropna(
    subset=["infrastructure_regime", mobility_intensity_var, nightlight_field]
).copy()
response_df["infrastructure_regime"] = response_df["infrastructure_regime"].astype(int)

regime_models = {}
regime_diagnostics = []

for regime_id in sorted(response_df["infrastructure_regime"].unique()):
    regime_subset = response_df.loc[response_df["infrastructure_regime"] == regime_id]

    model_result = fit_regime_spline(
        x=regime_subset[mobility_intensity_var].values,
        y=regime_subset[nightlight_field].values,
        n_knots=n_knots,
        degree=spline_degree,
        bootstrap_iterations=bootstrap_iterations,
        grid_points=grid_points,
        random_state=response_random_state + regime_id,
    )

    regime_models[regime_id] = model_result

    regime_diagnostics.append(
        {
            "regime": regime_id,
            "observations_used": model_result["n_used"],
            "elasticity_p10": round(model_result["elasticity_p10"], 6),
            "elasticity_mean": round(model_result["elasticity_mean"], 6),
            "elasticity_p90": round(model_result["elasticity_p90"], 6),
            "p90_band_width": round(model_result["p90_band_width"], 6),
            "max_abs_curvature": round(model_result["max_abs_curvature"], 6),
        }
    )

regime_diagnostics_df = pd.DataFrame(regime_diagnostics)

print_banner("REGIME-SPECIFIC RESPONSE DIAGNOSTICS")
display(regime_diagnostics_df)
```

```python
# ============================================================
# SECTION 5.4 — Plot regime-specific mobility–economic response curves
# ============================================================

plt.figure(figsize=(10, 6.5))

for regime_id, model_result in regime_models.items():
    plt.plot(
        model_result["x_grid"],
        model_result["y_pred"],
        linewidth=2,
        label=f"Regime {regime_id}",
    )
    plt.fill_between(
        model_result["x_grid"],
        model_result["pred_low"],
        model_result["pred_high"],
        alpha=0.15,
    )

plt.xlabel("Mobility Intensity (standardized log visit volume)")
plt.ylabel("Mean Nighttime Light Intensity")
plt.title("Regime-Specific Nonlinear Mobility–Economic Relationships", fontsize=13, pad=12)
plt.legend(title="Infrastructure regime")
plt.tight_layout()
plt.show()
```

The response curves showed whether the economic meaning of mobility intensity changed across structurally distinct mobility environments. In other words, this step evaluated whether similar levels of mobility intensity aligned differently with local economic intensity depending on the regime in which they occurred.

The economic validation produced two complementary results. First, nighttime luminosity differed systematically across the recovered mobility infrastructure regimes. Second, the geometry of the mobility–economic relationship also varied across those regimes, indicating that mobility intensity did not map onto local economic intensity in the same way everywhere.

This result strengthened the interpretation of the framework. The recovered regimes were not merely statistical partitions of mobility covariance structure. They also corresponded to distinct economic environments and distinct mobility–economic response profiles.

# 6. Multiscale Reconstruction

The previous sections showed that mobility-based infrastructure formed a spatially differentiated field across the national territory. But tract-level analysis still left an important question open: how much spatial variation remained hidden **within** those administrative units?

To address that question, we implemented a **mass-preserving downscaling** exercise in a São Paulo metropolitan window. The goal was not to change tract-level values, but to redistribute them across a fine spatial grid while preserving aggregate tract-level consistency.

This step mattered because census tracts could still conceal substantial intra-tract heterogeneity. A single tract-level value might summarize very different local conditions, especially in dense metropolitan environments where commercial corridors, transportation hubs, and residential areas coexisted within the same administrative unit.

```python
# ============================================================
# SECTION 6 — Imports and configuration
# ============================================================

from contextlib import ExitStack

from shapely.geometry import box
from rasterio.features import rasterize
from rasterio.merge import merge
from rasterio.transform import from_origin, array_bounds, rowcol
from rasterio.warp import reproject, Resampling
from matplotlib.colors import Normalize, TwoSlopeNorm
from matplotlib.cm import ScalarMappable

# Spatial downscaling configuration
sp_bbox = (-47.2, -24.2, -45.8, -22.8)   # São Paulo metropolitan window (lon_min, lat_min, lon_max, lat_max)
downscale_crs = 31983                    # SIRGAS 2000 / UTM zone 23S
grid_resolution_m = 100                  # 100-meter grid
nightlight_weight_floor = 0.0            # only positive VIIRS values contribute to weighted redistribution
focus_buffer_m = 1500                    # buffer around most fragmented tract for local visualization
```

### 6.1 Selecting the metropolitan study area

For the downscaling exercise, we focused on a São Paulo metropolitan window that captured a dense and heterogeneous urban environment suitable for evaluating intra-tract variation. We preserved the tract-level analytical fields from the national GeoPackage and prepared the selected tracts in both geographic and metric coordinate systems.

```python
# ============================================================
# SECTION 6.1 — Select São Paulo metropolitan window
# ============================================================

gdf_geo = gdf.to_crs(epsg=4326).copy()

lon_min, lat_min, lon_max, lat_max = sp_bbox
sp_geo = gdf_geo.cx[lon_min:lon_max, lat_min:lat_max].copy()
sp_metric = sp_geo.to_crs(epsg=downscale_crs).copy().reset_index(drop=True)

if len(sp_metric) == 0:
    raise ValueError("The São Paulo metropolitan window returned no census tracts. Please verify the bounding box.")

sp_metric["tract_code"] = np.arange(1, len(sp_metric) + 1)

sp_summary = pd.DataFrame(
    {
        "metric": [
            "selected_census_tracts",
            "bounding_box_lon_lat",
            "metric_crs",
            "grid_resolution_m",
        ],
        "value": [
            len(sp_metric),
            str(sp_bbox),
            f"EPSG:{downscale_crs}",
            grid_resolution_m,
        ],
    }
)

print_banner("SÃO PAULO METROPOLITAN WINDOW")
display(sp_summary)
```

### 6.2 Preparing the ancillary VIIRS surface at 100-meter resolution

The paper’s downscaling logic used VIIRS nighttime light intensity as the ancillary layer guiding within-tract redistribution. To preserve that logic while keeping the notebook efficient, we selected only the VIIRS tiles intersecting the São Paulo window, merged them locally, and reprojected the resulting luminosity surface to the same 100-meter grid used for downscaling.

```python
# ============================================================
# SECTION 6.2 — Select VIIRS tiles intersecting the study area
# ============================================================

study_bbox_geom = box(*sp_metric.to_crs(epsg=4326).total_bounds)

intersecting_viirs_tiles = []

for tile_path in viirs_tiles:
    with rasterio.open(tile_path) as src:
        tile_geom = box(*src.bounds)
        if tile_geom.intersects(study_bbox_geom):
            intersecting_viirs_tiles.append(tile_path)

if len(intersecting_viirs_tiles) == 0:
    raise ValueError("No VIIRS tiles intersect the São Paulo study window.")

print_banner("VIIRS TILES USED FOR DOWNSCALING")
print(f"Intersecting VIIRS tiles: {len(intersecting_viirs_tiles)}")
for tile in intersecting_viirs_tiles:
    print("-", tile.name)
```

```python
# ============================================================
# SECTION 6.2 — Build 100m target grid and reproject VIIRS
# ============================================================

xmin, ymin, xmax, ymax = sp_metric.total_bounds
width = int(np.ceil((xmax - xmin) / grid_resolution_m))
height = int(np.ceil((ymax - ymin) / grid_resolution_m))

target_transform = from_origin(xmin, ymax, grid_resolution_m, grid_resolution_m)
target_extent = array_bounds(height, width, target_transform)  # west, south, east, north

with ExitStack() as stack:
    src_datasets = [stack.enter_context(rasterio.open(tile)) for tile in intersecting_viirs_tiles]

    merged_viirs, merged_transform = merge(
        src_datasets,
        bounds=sp_metric.to_crs(epsg=4326).total_bounds,
        nodata=-1,
    )

    merged_viirs = merged_viirs[0]  # single-band output
    viirs_100m = np.full((height, width), np.nan, dtype="float32")

    reproject(
        source=merged_viirs,
        destination=viirs_100m,
        src_transform=merged_transform,
        src_crs=src_datasets[0].crs,
        dst_transform=target_transform,
        dst_crs=f"EPSG:{downscale_crs}",
        src_nodata=-1,
        dst_nodata=np.nan,
        resampling=Resampling.bilinear,
    )

viirs_100m = np.where(np.isfinite(viirs_100m) & (viirs_100m > nightlight_weight_floor), viirs_100m, 0.0)

grid_summary = pd.DataFrame(
    {
        "metric": [
            "grid_width_pixels",
            "grid_height_pixels",
            "total_grid_cells",
            "positive_viirs_cells",
        ],
        "value": [
            width,
            height,
            width * height,
            int((viirs_100m > 0).sum()),
        ],
    }
)

print_banner("100M TARGET GRID AND ANCILLARY SURFACE")
display(grid_summary)
```

### 6.3 Mass-preserving tract-to-grid redistribution

We then assigned each 100-meter grid cell to a census tract and redistributed tract-level MII values across the grid using normalized VIIRS-derived weights.

For each tract, the within-tract weights summed to one. This ensured exact mass preservation: the sum of the redistributed cell values within a tract remained equal to the original tract-level MII. When a tract contained no positive ancillary signal, we fell back to a uniform allocation rule within that tract.

```python
# ============================================================
# SECTION 6.3 — Rasterize tracts and compute mass-preserving allocation
# ============================================================

# Rasterize tract membership onto the same 100m grid
tract_raster = rasterize(
    shapes=zip(sp_metric.geometry, sp_metric["tract_code"]),
    out_shape=(height, width),
    fill=0,
    transform=target_transform,
    all_touched=False,
    dtype="int32",
)

valid_cell_mask = tract_raster > 0
tract_codes_flat = tract_raster[valid_cell_mask].astype(np.int32)
viirs_weights_flat = viirs_100m[valid_cell_mask].astype(np.float64)

# Per-tract lookup arrays
max_code = int(sp_metric["tract_code"].max())
pixel_counts = np.bincount(tract_codes_flat, minlength=max_code + 1)
weight_sums = np.bincount(tract_codes_flat, weights=viirs_weights_flat, minlength=max_code + 1)

mii_by_code = np.zeros(max_code + 1, dtype=np.float64)
mii_by_code[sp_metric["tract_code"].values] = sp_metric["mii"].values

mii_norm_by_code = np.zeros(max_code + 1, dtype=np.float64)
mii_norm_by_code[sp_metric["tract_code"].values] = sp_metric["mii_norm"].values

sum_lookup = weight_sums[tract_codes_flat]
count_lookup = pixel_counts[tract_codes_flat]

normalized_weights_flat = np.empty_like(viirs_weights_flat, dtype=np.float64)

use_viirs_mask = sum_lookup > 0
use_uniform_mask = ~use_viirs_mask

normalized_weights_flat[use_viirs_mask] = viirs_weights_flat[use_viirs_mask] / sum_lookup[use_viirs_mask]
normalized_weights_flat[use_uniform_mask] = 1.0 / count_lookup[use_uniform_mask]

# Mass-preserving downscaled MII allocation
mii_100m_flat = mii_by_code[tract_codes_flat] * normalized_weights_flat

# Relative intensity versus a uniform within-tract allocation
# Under uniform allocation, each cell would receive 1 / n_i of tract mass.
relative_intensity_flat = normalized_weights_flat * count_lookup

# Rebuild raster layers
mii_100m = np.full((height, width), np.nan, dtype=np.float64)
relative_intensity = np.full((height, width), np.nan, dtype=np.float64)
tract_code_raster = np.full((height, width), 0, dtype=np.int32)

mii_100m[valid_cell_mask] = mii_100m_flat
relative_intensity[valid_cell_mask] = relative_intensity_flat
tract_code_raster[valid_cell_mask] = tract_codes_flat

# Exact mass-preservation check
allocated_sum_by_tract = np.bincount(tract_codes_flat, weights=mii_100m_flat, minlength=max_code + 1)
original_sum_by_tract = mii_by_code.copy()
mass_error = allocated_sum_by_tract[1:] - original_sum_by_tract[1:]

mass_check = pd.DataFrame(
    {
        "metric": [
            "tracts_downscaled",
            "assigned_grid_cells",
            "max_absolute_mass_error",
            "mean_absolute_mass_error",
            "tracts_using_uniform_fallback",
        ],
        "value": [
            len(sp_metric),
            int(valid_cell_mask.sum()),
            float(np.max(np.abs(mass_error))),
            float(np.mean(np.abs(mass_error))),
            int((weight_sums[1:] <= 0).sum()),
        ],
    }
)

print_banner("MASS-PRESERVING DOWNSCALING CHECK")
display(mass_check)
```

The preservation check verified the core algebraic constraint of the downscaling operator: tract-level values were redistributed across the 100-meter grid without altering tract-level totals. The right-hand surface therefore represented within-tract redistribution, not the creation of new aggregate mass.

### 6.4 Comparing tract-level structure and the downscaled surface

We next compared the original tract-level infrastructure field with the reconstructed fine-scale surface. The tract map preserved the macro-scale structure of the MII, whereas the 100-meter surface revealed where infrastructure mass was concentrated or attenuated relative to a uniform within-tract allocation.

```python
# ============================================================
# SECTION 6.4 — Visual comparison: tract-level vs downscaled surface
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(15, 7.5))

# Left panel — tract-level normalized MII
sp_metric.plot(
    column="mii_norm",
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    linewidth=0.2,
    edgecolor="black",
    ax=axes[0],
)

axes[0].set_title("Tract-Level Infrastructure (Normalized MII)", fontsize=13, pad=12)
axes[0].axis("off")

# Right panel — relative intensity surface
valid_relative = relative_intensity[np.isfinite(relative_intensity)]
q05 = np.nanpercentile(valid_relative, 5)
q95 = np.nanpercentile(valid_relative, 95)

if q05 < 1.0 < q95:
    norm_right = TwoSlopeNorm(vmin=q05, vcenter=1.0, vmax=q95)
else:
    norm_right = Normalize(vmin=q05, vmax=q95)

west, south, east, north = target_extent
im = axes[1].imshow(
    relative_intensity,
    extent=[west, east, south, north],
    origin="upper",
    cmap="magma",
    norm=norm_right,
)

sp_metric.boundary.plot(
    ax=axes[1],
    linewidth=0.1,
    edgecolor="black",
)

axes[1].set_title("100m Relative Intensity Surface", fontsize=13, pad=12)
axes[1].axis("off")

# Colorbars
sm_left = ScalarMappable(norm=Normalize(vmin=-1, vmax=1), cmap="coolwarm")
sm_left.set_array([])

cbar_left = fig.colorbar(sm_left, ax=axes[0], fraction=0.046, pad=0.04)
cbar_left.set_label("Normalized MII")

cbar_right = fig.colorbar(im, ax=axes[1], fraction=0.046, pad=0.04)
cbar_right.set_label("Relative intensity vs. uniform allocation")

plt.tight_layout()
plt.show()
```

The reconstructed surface revealed where tract-level infrastructure mass concentrated more strongly and where it attenuated relative to a uniform within-tract allocation. This made intra-tract heterogeneity visible without changing the aggregate tract-level structure recovered earlier in the notebook.

### 6.5 Intra-tract fragmentation diagnostics

Finally, we quantified intra-tract fragmentation by measuring how unevenly the downscaled allocation varied within each tract. If tracts were internally homogeneous, their relative intensity would remain close to one throughout the grid. Greater dispersion implied stronger sub-tract differentiation.

```python
# ============================================================
# SECTION 6.5 — Fragmentation diagnostics by tract
# ============================================================

fragmentation_df = pd.DataFrame(
    {
        "tract_code": tract_codes_flat,
        "relative_intensity": relative_intensity_flat,
    }
)

tract_fragmentation = (
    fragmentation_df
    .groupby("tract_code")["relative_intensity"]
    .agg(["count", "mean", "std", "min", "max"])
    .reset_index()
    .rename(columns={"std": "fragmentation_index"})
)

tract_fragmentation["fragmentation_index"] = tract_fragmentation["fragmentation_index"].fillna(0.0)

sp_metric = sp_metric.merge(
    tract_fragmentation[["tract_code", "count", "fragmentation_index", "min", "max"]],
    on="tract_code",
    how="left",
)

fragmentation_summary = pd.DataFrame(
    {
        "metric": [
            "tracts_evaluated",
            "median_fragmentation_index",
            "mean_fragmentation_index",
            "95th_percentile_fragmentation_index",
            "max_fragmentation_index",
        ],
        "value": [
            len(sp_metric),
            round(sp_metric["fragmentation_index"].median(), 6),
            round(sp_metric["fragmentation_index"].mean(), 6),
            round(sp_metric["fragmentation_index"].quantile(0.95), 6),
            round(sp_metric["fragmentation_index"].max(), 6),
        ],
    }
)

print_banner("INTRA-TRACT FRAGMENTATION SUMMARY")
display(fragmentation_summary)

top_fragmented = (
    sp_metric[["tract_code", "ct_id", "fragmentation_index", "count", "mii", "mii_norm"]]
    .sort_values("fragmentation_index", ascending=False)
    .head(10)
    .reset_index(drop=True)
)

print_banner("MOST FRAGMENTED TRACTS")
display(top_fragmented)
```

```python
# ============================================================
# SECTION 6.5 — Local view of the most fragmented tract
# ============================================================

most_fragmented_tract = sp_metric.sort_values("fragmentation_index", ascending=False).iloc[0]
focus_geom = most_fragmented_tract.geometry.buffer(focus_buffer_m)
fxmin, fymin, fxmax, fymax = focus_geom.bounds

# Convert map window to raster row/column slices
row_min, col_min = rowcol(target_transform, fxmin, fymax)
row_max, col_max = rowcol(target_transform, fxmax, fymin)

row_min = max(0, min(height - 1, row_min))
row_max = max(0, min(height, row_max))
col_min = max(0, min(width - 1, col_min))
col_max = max(0, min(width, col_max))

sub_relative = relative_intensity[row_min:row_max, col_min:col_max]

x_left = xmin + col_min * grid_resolution_m
x_right = xmin + col_max * grid_resolution_m
y_top = ymax - row_min * grid_resolution_m
y_bottom = ymax - row_max * grid_resolution_m

fig, ax = plt.subplots(figsize=(7, 7))

im = ax.imshow(
    sub_relative,
    extent=[x_left, x_right, y_bottom, y_top],
    origin="upper",
    cmap="magma",
    norm=norm_right,
)

sp_metric.iloc[[most_fragmented_tract.name]].boundary.plot(
    ax=ax,
    linewidth=1.5,
    edgecolor="cyan",
)

ax.set_title("Most Fragmented Tract: Local 100m Relative Intensity", fontsize=13, pad=12)
ax.axis("off")

cbar = fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
cbar.set_label("Relative intensity vs. uniform allocation")

plt.tight_layout()
plt.show()
```

The fragmentation diagnostics showed that many census tracts concealed substantial internal differentiation. Even when a tract appeared homogeneous in the tract-level surface, the downscaled reconstruction often revealed localized pockets of concentration and attenuation within the same administrative unit.

This result completed the multiscale logic of the framework. We first recovered a national latent infrastructure field, then identified spatially differentiated regimes, then validated those regimes economically, and finally showed that tract-level representation still masked important intra-tract variation.

### 7.3 Limitations

The notebook also had clear limitations.

**First, the mobility data were proprietary and pre-aggregated.**  
Although the shared GeoPackage made the workflow reproducible at the analytical stage, exact replication with the original provider data required access to the same commercial source.

**Second, the framework was structural rather than causal.**  
The analyses identified mobility-based infrastructure conditions associated with local economic intensity, but they did not estimate causal effects or isolate exogenous variation in local activity.

**Third, the empirical demonstration used a single national setting and a limited temporal window.**  
The mobility signals reflected a tract-level snapshot, and the economic validation relied on annual nighttime light intensity for the same broad period. A richer temporal design could examine how mobility-based infrastructure evolved over time.

**Fourth, the downscaling exercise was intentionally framed as a pilot multiscale reconstruction.**  
Its role was to reveal concealed intra-tract variation while preserving tract-level consistency, not to claim that the 100-meter surface directly observed ground-truth micro-scale market outcomes.

These limitations did not weaken the value of the framework. Instead, they clarified its current scope and pointed to the next steps for extension and refinement.
